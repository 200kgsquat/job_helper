## Cleaning from stop words

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def remove_stopwords(text):
    return " ".join([word for word in text.split() if word.lower() not in ENGLISH_STOP_WORDS])

df['description'] = df['description'].apply(remove_stopwords)
```

## Lowercasting

```python
df['description'] = df['description'].str.lower()
```

## Remove extra spaces

```python
df['description_clean'] = df['description'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

## Remove special charecters

```python
def clean_spaces_around_punct(text):
    # Remove spaces before punctuation
    text = re.sub(r'\s+([.,!?;:])', r'\1', text)
    # Ensure exactly one space after punctuation (if not end of line)
    text = re.sub(r'([.,!?;:])([^\s])', r'\1 \2', text)
    return text.strip()

df['description'] = df['description'].apply(clean_spaces_around_punct)
```

## Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer


lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])

# Apply to your DataFrame
df['description'] = df['description'].apply(lemmatize_text)
```

# Handle punctuation

```python
import re

# Clean description: lowercase and remove non-word characters (punctuation)
df['description'] = (
    df['description']
    .fillna('')
    .astype(str)
    .apply(lambda x: ' '.join(re.findall(r'\b\w+\b', x.lower())))
)

print(df['description'].head())
```

## Handle empte stings

```python
# Ensure descriptions are strings and strip extra spaces
df['description'] = df['description'].fillna('').astype(str).str.strip()

# Remove empty descriptions
df = df[df['description'].str.len() > 0]

# Check the number of examples
print(f"Number of examples after cleaning: {len(df)}")
```
